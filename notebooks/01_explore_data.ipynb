# Explore SSDD
Sanity-check the dataset loader and visualize a few labeled SAR chips before training.

```python
import sys
sys.path.append('../src')

from dataset import SSDDDataset
import matplotlib.pyplot as plt
import matplotlib.patches as patches

dataset = SSDDDataset('../data')
print(f'{len(dataset)} images')
```

```python
image, target = dataset[0]
fig, ax = plt.subplots(1, figsize=(6, 6))
ax.imshow(image)
for box in target['boxes']:
    x0, y0, x1, y1 = box.tolist()
    ax.add_patch(patches.Rectangle((x0, y0), x1 - x0, y1 - y0, linewidth=2, edgecolor='lime', facecolor='none'))
ax.axis('off')
plt.show()
```
